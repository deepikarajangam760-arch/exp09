```python
#Step 4: Display the original image.
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('zebra.jpeg')

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(5,5))
plt.imshow(img)
plt.title('Original Image')
plt.axis('off')
plt.show()
```

# STEP 6: IMAGE EROSION

Apply the erosion operation using the created kernel. Remove pixels from the boundaries of foreground objects. Display the eroded image.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Define kernel
kernel = np.ones((5,5), np.uint8)

# Apply erosion
erosion = cv2.erode(img, kernel, iterations=1)

plt.imshow(erosion, cmap='gray')
plt.title("Erosion Image")
plt.axis('off')
plt.show()
```

# STEP 7: Image Dilation

Apply the dilation operation using the same kernel. Add pixels to the boundaries of foreground objects. Display the dilated image.

```python
dilation = cv2.dilate(img, kernel, iterations=1)
plt.imshow(dilation)
plt.title("Dilation Image")
plt.axis('off')
plt.show()
```

```python
# Step 8: Compare the original, eroded, and dilated images.
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(erosion)
plt.title("Erosion")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(dilation)
plt.title("Dilation")
plt.axis('off')

plt.tight_layout()
plt.show()
```
